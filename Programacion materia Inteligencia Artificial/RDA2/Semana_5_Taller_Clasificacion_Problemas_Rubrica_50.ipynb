## Recordatorio conceptual

| Tipo de problema | Característica principal | Técnica común |
|---|---|---|
| Supervisado | Existe una variable objetivo que se quiere predecir | Clasificación o regresión |
| No supervisado | No existe variable objetivo; se buscan grupos o patrones | Clustering o reducción de dimensionalidad |
| Descriptivo | Solo se quiere explorar, resumir o comprender los datos | EDA / estadística descriptiva |

---

## Regla básica

> Si hay una columna que quiero predecir, probablemente es supervisado.  
> Si no hay target y quiero encontrar grupos, probablemente es no supervisado.  
> Si solo quiero resumir o explorar datos, es descriptivo.

## Conclusión orientativa

La selección de técnica depende del problema y no del gusto del analista. Si existe una variable objetivo y se quiere predecir una categoría, el problema es supervisado de clasificación. Si la variable objetivo es numérica continua, se trata de regresión. Si no existe target y se buscan grupos o estructuras ocultas, corresponde a un enfoque no supervisado. Si solo se desea resumir, contar o explorar datos, el análisis es descriptivo o exploratorio. Antes de entrenar un modelo, siempre se debe definir el problema, revisar los datos, hacer EDA e identificar si existe una variable objetivo.
