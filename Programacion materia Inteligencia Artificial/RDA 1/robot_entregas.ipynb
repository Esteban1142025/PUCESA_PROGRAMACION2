```python
from collections import deque

# --- Implementación de BFS para encontrar un camino ---
def find_path_bfs(graph, start, goal):
    # La cola almacenará listas, donde cada lista es un camino.
    queue = deque([ [start] ])
    visited = set([start])

    while queue:
        # Obtiene el primer camino de la cola.
        path = queue.popleft()
        node = path[-1] # El último nodo del camino es el actual.

        if node == goal:
            return path # Se encontro un camino.

        # Explora los vecinos del nodo actual.
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                new_path = list(path)
                new_path.append(neighbor)
                queue.append(new_path)
    
    return None # Si la cola se vacía, no hay camino.

# --- Implementación de DFS para encontrar un camino (versión recursiva) ---
def find_path_dfs(graph, start, goal, path=None):
    if path is None:
        path = [start]

    if start == goal:
        return path # Caso base: hemos llegado al destino.
    
    for neighbor in graph[start]:
        if neighbor not in path: # Evita ciclos simples
            new_path = find_path_dfs(graph, neighbor, goal, path + [neighbor])
            if new_path:
                return new_path # Propaga la solución encontrada.

    return None # No se encontró camino por esta rama.

# --- Grafo del mapa de la ciudad ---
mapa_ciudad = {
'Almacen': ['Parque', 'Oficina de Correos'],
'Parque': ['Centro Comercial'],
'Oficina de Correos': ['Mercado'],
'Centro Comercial': ['Mercado', 'Barrio Residencial'],
'Mercado': ['Casa del Cliente'],
'Barrio Residencial': ['Casa del Cliente'],
'Casa del Cliente': []
}

# --- Ejecución del Ejercicio ---
start_node = 'Almacen'
goal_node = 'Casa del Cliente'

print("--- Búsqueda BFS ---")
path_bfs = find_path_bfs(mapa_ciudad, start_node, goal_node)
print(f"Camino encontrado: {path_bfs}")

print("\n--- Búsqueda DFS ---")
path_dfs = find_path_dfs(mapa_ciudad, start_node, goal_node)
print(f"Camino encontrado: {path_dfs}")
```

La ruta encontrada fue la siguiente: ['Almacen', 'Oficina de Correos', 'Mercado', 'Casa del Cliente'] y necesitó dar 3 pasos

¿Qué ruta encontró DFS? ¿Es diferente a la de BFS?\
La ruta encontrada fue la siguiente: ['Almacen', 'Parque', 'Centro Comercial', 'Mercado', 'Casa del Cliente']. Si es diferente ya que BFS toma la ruta por la oficina de correos en lugar del parque como lo realiza DFS.

¿Por qué crees que encontraron rutas diferentes (o la misma ruta)?\
Por el lado de BFS, este se encarga de explorar todos los nodos vecinos nivel por nivel tratando de buscar siempre la ruta mas corta. Mientras que DFS realiza una exploracion mas profunda buscando la primera opcion entre los vecinos antes de retroceder y buscar nuevamente hasta llegar al objetivo.
