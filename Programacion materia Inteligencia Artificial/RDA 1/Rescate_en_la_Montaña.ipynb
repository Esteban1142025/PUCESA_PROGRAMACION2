```python
# Grafo con tiempos de viaje en minutos.z
mapa_montana = {
'Campamento Base': {'Paso Rocoso': 30, 'Puente Colgante': 20},
'Paso Rocoso': {'Cascada': 60},
'Puente Colgante': {'Cascada': 40, 'Cueva': 50},
'Cascada': {'Refugio': 70},
'Cueva': {'Refugio': 60},
'Refugio': {}
}

# Tiempo estimado hasta el Refugio
heuristica_tiempo = {
'Campamento Base': 100, 'Paso Rocoso': 60,
'Puente Colgante': 80, 'Cascada': 50,
'Cueva': 40, 'Refugio': 0
}
```

```python
import heapq

# --- Implementación de Búsqueda Voraz Primero el Mejor ---
def greedy_best_first(graph, start, goal, heuristics):
    # La cola de prioridad ordena los nodos solo por su valor heurístico.
    priority_queue = [(heuristics[start], [start])] # (h(n), camino)
    visited = set()

    while priority_queue:
        h, path = heapq.heappop(priority_queue)
        node = path[-1]

        if node in visited: continue
        visited.add(node)
        
        if node == goal: return path

        for neighbor in graph[node]:
            if neighbor not in visited:
                heapq.heappush(priority_queue, (heuristics[neighbor], path + [neighbor]))
    return None

# --- Implementación de Búsqueda A* (A-Estrella) ---
def a_star_search(graph, start, goal, heuristics):
    # f(n) = g(n) + h(n)
    # g(n) es el costo real del camino desde el inicio.
    # h(n) es el costo estimado hasta el final.
    priority_queue = [(0 + heuristics[start], 0, [start])] # (f(n), g(n), camino)
    visited = set()

    while priority_queue:
        f, g, path = heapq.heappop(priority_queue)
        node = path[-1]

        if node in visited: continue
        visited.add(node)

        if node == goal: return path, g

        for neighbor, cost in graph[node].items():
            if neighbor not in visited:
                new_g = g + cost
                new_f = new_g + heuristics[neighbor]
                heapq.heappush(priority_queue, (new_f, new_g, path + [neighbor]))
    return None, -1
```

```python
# --- Ejecución y resultados ---
inicio = 'Campamento Base'
objetivo = 'Refugio'

# Greedy Best-First Search
camino_greedy = greedy_best_first(mapa_montana, inicio, objetivo, heuristica_tiempo)
print("--- Búsqueda Voraz Primero el Mejor (Greedy Best-First) ---")
print(f"Camino escogido: {' -> '.join(camino_greedy)}")

# A* Search
camino_astar, costo_astar = a_star_search(mapa_montana, inicio, objetivo, heuristica_tiempo)
print("\n--- Búsqueda A* (A-Estrella) ---")
print(f"Camino escogido: {' -> '.join(camino_astar)}")
print(f"Costo total (minutos): {costo_astar}")
```

**¿Qué ruta encontró el algoritmo Greedy Best-First y cuál fue su costo total en minutos?**\
**Ruta encontrada:** Campamento Base -> Paso Rocoso -> Cascada -> Refugio\
**Costo total:** 160 minutos el cual se calculo como 30 min hasta Paso Rocoso + 60 min hasta Cascada + 70 min hasta Refugio\

**¿Qué ruta encontró el algoritmo A(*) y cuál fue su costo total?**\
**Ruta encontrada:** Campamento Base -> Puente Colgante -> Cascada -> Refugio (o también puede ser a través de la Cueva, ya que ambas suman el mismo valor óptimo)\
**Costo total:** 130 minutos (calculado como 20 min hasta Puente Colgante + 40 min hasta Cascada + 70 min hasta Refugio)\

**¿Por qué el algoritmo voraz se equivocó (si lo hizo)? ¿Qué información ignoró?**\
**Por qué se equivocó:** Tomó una mala decisión en el primer paso al elegir Paso Rocoso en lugar de Puente Colgante, porque creía que estaba más cerca de la meta\
**Información que ignoró:** Ignoró por completo el costo real acumulado del camino recorrido. El algoritmo voraz solo mira la heurística, prefiriendo un nodo con una estimación más baja al futuro sin importarle si el tramo que acaba de caminar le costó demasiado tiempo\

**En una situación de rescate real, ¿por qué es CRUCIAL usar un algoritmo óptimo como A(*) en lugar de uno simplemente voraz?**\
**Optimización de recursos:** En un rescate, el tiempo, la energía de los rescatistas, el combustible y las reservas de oxígeno son limitados. Un algoritmo voraz puede guiar al equipo por senderos que parecen cortos sobre el mapa pero que en la práctica acumulan retrasos o peligros fatales.\
**Eficiencia:** A* equilibra de forma inteligente el esfuerzo ya realizado con la distancia restante, asegurando encontrar el camino verdaderamente más rápido y seguro, lo cual puede marcar la diferencia entre la vida y la muerte en una emergencia.
